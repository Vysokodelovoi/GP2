```python
from tqdm import trange
import openai
import ast
import json
result = []
def form_prompt(review):
    return f'''
        Тебе необходимо выделить в данном ревью данные для анализа. Вот шаблон, по которому тебе необходимо отвечать:
        {json_format}, строго соответствуй этому формату
        Вот отзыв который тебе необходимо обработать: {review}. Answer with only json object.
    '''
YANDEX_CLOUD_MODEL = "gemma-3-27b-it/latest"
json_format = '''
{
    sentiment: positive | negative, # : Отзыв является негативным или позитивным
    reason_n: Обслуживание персонала | Качество еды | Цены | Ассортимент | Качество доставки | Чистота и атмосфера в заведении | None | Список причин [reason1, reason2, ...] # : По какой причине не понравилась пиццерия, ИСПОЛЬЗУЙ ТОЛЬКО ЭТИ КАТЕГОРИИ
    reason_p: Обслуживание персонала | Качество еды | Цены | Ассортимент | Качество доставки | Чистота и атмосфера в заведении | None | Список причин [reason1, reason2, ...] # : По какой причине понравилась пиццерия, ИСПОЛЬЗУЙ ТОЛЬКО ЭТИ КАТЕГОРИИ
}
'''
YANDEX_CLOUD_FOLDER = "b1g8qv08h6hjp53m3ota"
YANDEX_CLOUD_API_KEY = "***"
client = openai.OpenAI(
    api_key=YANDEX_CLOUD_API_KEY,
    base_url="https://rest-assistant.api.cloud.yandex.net/v1",
    project=YANDEX_CLOUD_FOLDER
)
for i in trange(len(df)):
    # print(df.iloc[i]['text'])
    # print(df.iloc[i]['company'])
    response = client.responses.create(
        model=f"gpt://{YANDEX_CLOUD_FOLDER}/{YANDEX_CLOUD_MODEL}",
        temperature=0.1,
        instructions="You are special model, designed to help to analyze the reviews of pizzerias. Answer only in json format",
        input=form_prompt(df.iloc[i]['text']),
        max_output_tokens=500
    )
    try:
        obj = json.loads(response.output_text[7:-3])
        # obj = ast.literal_eval(response.output_text)
        result.append({'company': df.iloc[i]['company'], 'sentiment': obj['sentiment'], 'reason_n': obj['reason_n'], 'reason_p': obj['reason_p'],
                   'review_text': df.iloc[i]['text']})
    except Exception as e:
        print(e)
        result.append({'company': df.iloc[i]['company'],  'review_text': df.iloc[i]['text']})
        print('Response text: ' + response.output_text[7:-3])
        print('review_text:')
result
```
